#Python Basics Questions



1. What is Python, and why is it popular?
     - Python is a high-level, interpreted programming language known for its simplicity, readability, and versatility. It was created by Guido van Rossum and first released in 1991.
          * Easy to read
          * Beginner-friendly
          * Powerful enough for complex applications

2. What is an interpreter in Python?
     - An interpreter is a program that reads and executes code line by line.
       In Python's case, when you write a .py file and run it, the Python interpreter processes your code one instruction at a time, immediately executing it.

3. What are pre-defined keywords in Python?  
     - Pre-defined keywords in Python are reserved words that have a special meaning in the language.
     * if, else, — for conditional statements
     * for, while, break, — for loops

4. Can keywords be used as variable names?
     - No, keywords cannot be used as variable names in Python. Because keywords have a special, pre-defined meaning in Python's syntax.

5. What is mutability in Python?
     - In Python, mutability refers to whether an object's value can be changed after its been created.

6. Why are lists mutable, but tuples are immutable?
     - There is the reason why list are mutable and tuple are immutable.

     * **Lists store references to objects in a dynamic array.** Since you might need to add, remove, or change items, lists are mutable.
     * **Tuples store references in a fixed array.** Changing a tuple would mean making a new one anyway, so Python makes them immutable by design.

7. What is the difference between “==” and “is” operators in Python?
     - There is the difference between those.

     "=="	Equality operator	Whether the values of two objects are equal.

     "is"	Identity operator	Whether two variables point to the exact same object in memory.

8. What are logical operators in Python?
     - Logical operators are used to combine conditional statements (expressions that evaluate to True or False).
       They return a boolean result (True or False) based on the logic between the values or expressions you're comparing.

9. What is type casting in Python?
     - Type casting (or type conversion) is the process of converting the value of one data type to another.

     In Python, you can convert between types like:
     * int (integer)
     * float (decimal numbers)
     * str (string)
     * bool (boolean)

10. What is the difference between implicit and explicit type casting?
     - There is a difference between those.

     Implicit Type Casting
     * Python automatically converts one data type to another when needed.
     * Happens behind the scenes.
     * Only happens when it's safe and lossless (like int to float).

     Explicit Type Casting
     * The programmer manually converts one data type to another using casting functions.
     * Happens by explicitly writing conversion code.
     * Can convert between any compatible types, even if data might be lost (like float to int).

11. What is the purpose of conditional statements in Python?
     - Conditional statements allow you to control the flow of your program based on certain conditions. In other words, they let you execute specific blocks of code only if certain conditions are True.

     Purpose of conditional statements.
     * Decision Making
     * Flow Control
     * Handling Edge Cases
     * Optimizing Logic

12. How does the elif statement work?
     - he elif (short for "else if") statement allows you to test multiple conditions sequentially after an if statement. If the first if condition is False, Python will check the conditions in the elif block(s).
     
     * The first if condition that evaluates to True will have its corresponding block of code executed.
     * If none of the if or elif conditions are True, the else block will be executed (if it's present).

13. What is the difference between for and while loops?
     - There is following difference between these.

     "for" Loop
     * Purpose: Used to iterate over a sequence (like a list, tuple, string, or range).
     * When to Use: When you know how many times you want to repeat the block of code (i.e., you know the number of iterations in advance).
     * Structure: Iterates through a sequence or a specific range of values.

     "while" Loop
     * Purpose: Used for repeating a block of code while a condition is true.
     * When to Use: When you don’t know exactly how many times the loop will run in advance, but you know the condition that will terminate it.
     * Structure: Repeats the code as long as the specified condition is True.

14. Describe a scenario where a while loop is more suitable than a for loop.
    
    - Imagine you're building a program where you want to prompt the user for a specific input, but you don't know how many attempts it will take for the user to give the correct input. In this case, you would use a while loop because you want to keep looping as long as the condition (correct input) is not met.

    * A while loop is ideal when you need to repeat an action based on a condition being True (e.g., waiting for valid user input or checking for a specific condition to be met).

```python
#6.

def modify_arguments(num, num_list):
    # Modify the immutable integer
    num += 10
    print("Inside function, num (immutable):", num)

    # Modify the mutable list
    num_list.append(99)
    print("Inside function, num_list (mutable):", num_list)


# Original values
number = 5
numbers_list = [1, 2, 3]

# Call the function
modify_arguments(number, numbers_list)

# Check values after function call
print("\nAfter function call:")
print("number (immutable):", number)
print("numbers_list (mutable):", numbers_list)
```

```python
#10

# List of strings representing numbers
string_list = ["10", "20", "0", "35", ""]

# Convert string list elements to integers
int_list = []
for item in string_list:
    try:
        int_list.append(int(item))
    except ValueError:
        int_list.append(0)  # If conversion fails, append 0

print("List converted to integers:", int_list)

# Convert string list elements to floats
float_list = []
for item in string_list:
    try:
        float_list.append(float(item))
    except ValueError:
        float_list.append(0.0)

print("List converted to floats:", float_list)

# Convert string list elements to booleans
bool_list = [bool(item) for item in string_list]
print("List converted to booleans:", bool_list)
```

```python
#13

# Initialize the sum variable
sum_of_evens = 0

# Loop through numbers from 1 to 50
for num in range(1, 51):
    if num % 2 == 0:  # Check if the number is even
        sum_of_evens += num  # Add even number to the sum

# Print the result
print("Sum of all even numbers between 1 and 50 is:", sum_of_evens)
```

```python
#14

# Get input string from the user
user_input = input("Enter a string:")

# Initialize the reversed string as an empty string
reversed_string = ""

# Set the initial index to the last character of the string
index = len(user_input) - 1

# Use a while loop to traverse the string backwards
while index >= 0:
    reversed_string += user_input[index]  # Add each character to the reversed string
    index -= 1  # Move to the previous character

# Print the reversed string
print("Reversed string:", reversed_string)
```

```python
#15

# Get the number from the user
num = int(input("Enter a number to calculate its factorial: "))

# Initialize the result variable to 1 (since factorial of 0 is 1)
factorial = 1

# Check if the number is negative
if num < 0:
    print("Factorial is not defined for negative numbers.")
else:
    # Use a while loop to calculate the factorial
    while num > 1:
        factorial *= num  # Multiply the current factorial value by num
        num -= 1  # Decrement the number by 1

    # Print the result
    print("Factorial is:", factorial)
```
